```python
def get_heatmap_color(val):
    val = max(0, min(100, val))

    if val < 50:
        norm = val / 50
        r, g, b = 255, int(255 * norm), 0
    else:
        norm = (val - 50) / 50
        r, g, b = int(255 * (1 - norm)), 255, 0

    # Adaptive pastelization (less white near yellow)
    dist_from_mid = abs(val - 50) / 50   # 0 at yellow, 1 at extremes
    mix_white = 0.25 + 0.25 * dist_from_mid  # ∈ [0.25, 0.5]

    r = int(r * (1 - mix_white) + 255 * mix_white)
    g = int(g * (1 - mix_white) + 255 * mix_white)
    b = int(b * (1 - mix_white) + 255 * mix_white)

    return f"{r:02X}{g:02X}{b:02X}"



def colorize_word(word, vals):
    assert len(word) == len(vals)
    return "".join(
        rf"\BGcolor{{{get_heatmap_color(v).lower()}}}{{{c}}}"
        for c, v in zip(word, vals)
    )

# Define words and monotonic scores (0 = worst, 100 = best)
worst_vals = [0, 10, 20, 30, 40]
and_vals   = [50, 55, 60]
best_vals  = [70, 80, 90, 95]

latex_string = (
    colorize_word("worst", worst_vals)
    + r"\BGcolor{" + get_heatmap_color(50).lower() + r"}{ }"
    + colorize_word("and", and_vals)
    + r"\BGcolor{" + get_heatmap_color(50).lower() + r"}{ }"
    + colorize_word("best", best_vals)
)

print(latex_string)
```
